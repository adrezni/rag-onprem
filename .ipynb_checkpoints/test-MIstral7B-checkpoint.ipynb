```python
#http://nim-gpt-oss-20b-predictor.pw-demo.svc.cluster.local
#initilaize the client targeting the NIM model
client = OpenAI(
    base_url = "http://nim-gpt-oss-20b-predictor.pw-demo.svc.cluster.local/v1",
    api_key = os.environ['api_key']
)
```

```python

#call chat completions API
completion = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": "You are a concise engineering assistant."},
        {"role": "user", "content": "What are the benefits of NVIDIA NIM on Red Hat OpenShift AI?"}
    ],
    temperature=0.2,
    top_p=0.7,
    max_tokens=1024,
    stream=True
)
```

```python

#stream response tokens
for chunk in completion:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)        
```
